# Identifier
In Python, an identifier is a name used to identify a variable, function, class, module, or other object.
Rules for identifiers,
1. can only start with alphabet or _ (underscore)
2. followed by 0 or more digits
3. keyword can not be used as identifier

```python
_ = "Bin"
print(_)
W = "pen"
print(W)
T_1 = "owl"
print(T_1)
name_1 = "kia"
print(name_1)
# We can use name followed by digits but can't use the digits followed by name.
```
