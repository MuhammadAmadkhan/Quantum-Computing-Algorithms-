# **Simulation**
Using classical computers to model and predict the behavior of quantum systems by solving quantum Circuits.
# **AerSimulator**
The AerSimulator is a high-performance, local simulator included with Qiskit. You can think of it as a powerful software tool on your computer that precisely calculates how a quantum circuit would behave, without needing access to real quantum hardware

```python
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
qc = QuantumCircuit(2)   # Taking two qubit
qc.h(0)
qc.cx(1,0)
qc.measure_all()  #measuring all qubits if you didn't measure then it will be remain in superpostion then you can not simulate.
simulator = AerSimulator()
counts = simulator.run(qc).result().get_counts() # you are simulating the circuit that u initialize basically you are running the (QC), result that u expecting
# on the output and the counts u can mention and u can let it too it will consider 1024 counts.
qc.draw("mpl")  # Here i drew the circuit just to show the circuit how it looks like.
```

```python
plot_histogram(counts)
```

# Quantum Teleportation

```python
from qiskit import QuantumRegister, ClassicalRegister
from qiskit_aer import AerSimulator
from qiskit.circuit.library import UGate
from qiskit.visualization import plot_histogram, array_to_latex
from qiskit.result import marginal_distribution

secret = QuantumRegister(1 ,"O")  #n quantum computing, registers are like containers that hold information.
Alice  = QuantumRegister(1, "A")  #A quantum register is a collection of qubits that store quantum information.
Bob    = QuantumRegister(1, "B")  #Each qubit can be in a superposition of |0⟩ and |1⟩ simultaneously.
cr     = ClassicalRegister(3 , "c")  #A classical register is a collection of classical bits that store the results of measurements.
                                     #These bits are just like regular computer bits - either 0 or 1.

Teleportation = QuantumCircuit(secret, Alice, Bob, cr)
Teleportation.h(Alice)
Teleportation.cx(Alice, Bob)
Teleportation.barrier()
Teleportation.cx(secret, Alice)
Teleportation.h(secret)
Teleportation.barrier()
Teleportation.measure(secret, cr[0])
Teleportation.measure(Alice, cr[1])
with Teleportation.if_test((cr[1],True)):
  Teleportation.x(Bob)
with Teleportation.if_test((cr[0],True)):
  Teleportation.z(Bob)

Teleportation.draw("mpl")
```

```python
simulator = AerSimulator()   #Parentheses mean "execute/call this function right now.

job = simulator.run(Teleportation, shots=1024)  #The dot means "access something that belongs to this object

result = job.result()
counts = result.get_counts()  #  get_counts belongs to result
plot_histogram(counts)

#result.get_counts     # ← just POINTING at the function, not running it
#result.get_counts()   # ← actually RUNNING it and getting the output
```

```python
qubit = QuantumRegister(1, "Q")
ebit0 = QuantumRegister(1, "A")
ebit1 = QuantumRegister(1, "B")
a = ClassicalRegister(1, "a")
b = ClassicalRegister(1, "b")

protocol = QuantumCircuit(qubit, ebit0, ebit1, a, b)

# Prepare ebit used for teleportation
protocol.h(ebit0)
protocol.cx(ebit0, ebit1)
protocol.barrier()

# Alice's operations
protocol.cx(qubit, ebit0)
protocol.h(qubit)
protocol.barrier()

# Alice measures and sends classical bits to Bob
protocol.measure(ebit0, a)
protocol.measure(qubit, b)
protocol.barrier()

# Bob uses the classical bits to conditionally apply gates
with protocol.if_test((a, 1)):
    protocol.x(ebit1)
with protocol.if_test((b, 1)):
    protocol.z(ebit1)

display(protocol.draw(output="mpl"))
```
